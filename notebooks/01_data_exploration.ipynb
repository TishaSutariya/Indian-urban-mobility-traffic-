```python
from pathlib import Path
import pandas as pd

# Find the project folder from the notebook location
PROJECT_PATH = Path.cwd().parent

# If VS Code opened the notebook with the project folder as the working directory
if not (PROJECT_PATH / "data" / "processed").exists():
    PROJECT_PATH = Path.cwd()

DATA_PATH = PROJECT_PATH / "data" / "processed"

print("Project folder:", PROJECT_PATH)
print("Data folder:", DATA_PATH)
print("Data folder exists:", DATA_PATH.exists())
```

```python
# Load all cleaned datasets

a1 = pd.read_csv(DATA_PATH / "RA2019_A1_cleaned.csv")
a2 = pd.read_csv(DATA_PATH / "RA2019_A2_cleaned.csv")
a47 = pd.read_csv(DATA_PATH / "RA2019_A47_cleaned.csv")
a48 = pd.read_csv(DATA_PATH / "RA2019_A48_cleaned.csv")
a49 = pd.read_csv(DATA_PATH / "RA2019_A49_cleaned.csv")
a50 = pd.read_csv(DATA_PATH / "RA2019_A50_cleaned.csv")
a51 = pd.read_csv(DATA_PATH / "RA2019_A51_cleaned.csv")
a52 = pd.read_csv(DATA_PATH / "RA2019_A52_cleaned.csv")

print("All 8 datasets loaded successfully!")
```

```python
# Create a dictionary containing all datasets

datasets = {
    "A1 - India Accident Trend": a1,
    "A2 - State Accidents": a2,
    "A47 - Accident Causes": a47,
    "A48 - Road Features": a48,
    "A49 - Junction Types": a49,
    "A50 - Traffic Control": a50,
    "A51 - Black Spots": a51,
    "A52 - High Accident Locations": a52
}

print("Datasets dictionary created successfully!")

for name, df in datasets.items():
    print(f"{name}: {df.shape[0]} rows × {df.shape[1]} columns")
```

```python
for name, df in datasets.items():
    print("\n" + "=" * 60)
    print(name)
    print("=" * 60)

    missing = df.isnull().sum()
    missing = missing[missing > 0]

    if len(missing) == 0:
        print("No missing values")
    else:
        print(missing)
```

```python
# Show the first 5 rows of each dataset

for name, df in datasets.items():
    print("\n" + "=" * 80)
    print(name)
    print("=" * 80)
    display(df.head())
```

```python
import matplotlib.pyplot as plt

# Convert important A1 columns to numeric
a1["Years"] = pd.to_numeric(a1["Years"], errors="coerce")
a1["Total Number of Road Accidents (in numbers)"] = pd.to_numeric(
    a1["Total Number of Road Accidents (in numbers)"],
    errors="coerce"
)

# Remove rows where year or accidents are missing
trend = a1.dropna(
    subset=["Years", "Total Number of Road Accidents (in numbers)"]
)

# Plot accident trend
plt.figure(figsize=(10, 5))

plt.plot(
    trend["Years"],
    trend["Total Number of Road Accidents (in numbers)"],
    marker="o"
)

plt.title("India Road Accident Trend")
plt.xlabel("Year")
plt.ylabel("Number of Road Accidents")
plt.grid(True)

plt.show()
```

```python
# Convert 2019 accident column to numeric
accident_col = "State/UT-Wise Total Number of Road Accidents during 2019 - Numbers"

a2[accident_col] = pd.to_numeric(
    a2[accident_col],
    errors="coerce"
)

# Remove missing values
state_accidents = a2.dropna(
    subset=[accident_col]
).copy()

# Sort states by number of accidents
state_accidents = state_accidents.sort_values(
    accident_col,
    ascending=False
)

# Select top 10
top10 = state_accidents.head(10)

# Display top 10
display(
    top10[
        ["States/UTs", accident_col]
    ]
)
```

```python
# Remove the aggregate "Total" row
state_accidents = state_accidents[
    state_accidents["States/UTs"] != "Total"
]

# Select top 10 states/UTs
top10 = state_accidents.head(10)

# Create bar chart
plt.figure(figsize=(10, 6))

plt.bar(
    top10["States/UTs"],
    top10[accident_col]
)

plt.title("Top 10 States/UTs by Road Accidents in 2019")
plt.xlabel("State/UT")
plt.ylabel("Number of Road Accidents")

plt.xticks(rotation=45, ha="right")
plt.tight_layout()

plt.show()
```

```python
# Display the main accident-cause columns
cause_columns = [
    "States/UTs",
    "Over-Speeding - Number of Accidents - Number",
    "Drunken Driving/ Consumption of alcohol and drug - Number of Accidents",
    "Use of Mobile Phone - Number of Accidents"
]

display(a47[cause_columns].head(10))
```

```python
# Convert over-speeding accidents to numeric
overspeed_col = "Over-Speeding - Number of Accidents - Number"

a47[overspeed_col] = pd.to_numeric(
    a47[overspeed_col],
    errors="coerce"
)

# Remove missing values and sort
overspeed = a47.dropna(
    subset=[overspeed_col]
).copy()

overspeed = overspeed.sort_values(
    overspeed_col,
    ascending=False
)

# Select top 10 locations
top10_overspeed = overspeed.head(10)

# Display results
display(
    top10_overspeed[
        ["States/UTs", overspeed_col]
    ]
)
```

```python
# Remove aggregate Total row
overspeed = overspeed[
    overspeed["States/UTs"] != "Total"
]

# Select top 10 locations
top10_overspeed = overspeed.head(10)

# Create chart
plt.figure(figsize=(10, 6))

plt.bar(
    top10_overspeed["States/UTs"],
    top10_overspeed[overspeed_col]
)

plt.title("Top 10 Locations by Over-Speeding-Related Road Accidents in 2019")
plt.xlabel("Location / Traffic Unit")
plt.ylabel("Number of Accidents")

plt.xticks(rotation=45, ha="right")
plt.tight_layout()

plt.show()
```

```python
# Accident count columns for major causes
cause_columns = {
    "Over-Speeding": "Over-Speeding - Number of Accidents - Number",
    "Drunken Driving": "Drunken Driving/ Consumption of alcohol and drug - Number of Accidents",
    "Wrong Side": "Driving on Wrong Side - Number of Accidents",
    "Jumping Red Light": "Jumping Red Light - Number of Accidents",
    "Mobile Phone": "Use of Mobile Phone - Number of Accidents"
}

# Calculate total accidents for each cause
cause_totals = {}

for cause, column in cause_columns.items():
    if column in a47.columns:
        cause_totals[cause] = pd.to_numeric(
            a47[column],
            errors="coerce"
        ).sum()

cause_summary = pd.DataFrame(
    list(cause_totals.items()),
    columns=["Cause", "Total Accidents"]
)

cause_summary = cause_summary.sort_values(
    "Total Accidents",
    ascending=False
)

display(cause_summary)
```

```python
plt.figure(figsize=(9, 6))

plt.bar(
    cause_summary["Cause"],
    cause_summary["Total Accidents"]
)

plt.title("Reported Road Accidents by Major Cause Categories")
plt.xlabel("Cause")
plt.ylabel("Number of Reported Accidents")

plt.xticks(rotation=20)
plt.tight_layout()

plt.show()
```

```python
# Show road-feature accident columns
road_columns = [
    col for col in a48.columns
    if "Number of Accidents" in col
]

print("Road-feature accident columns:")
for col in road_columns:
    print(col)
```

```python
# Road-feature accident columns
road_features = {
    "Straight Road": "Straight Road - Number of Accidents - number",
    "Curved Road": "Curved Road - Number of Accidents",
    "Bridge": "Bridge - Number of Accidents",
    "Culvert": "Culvert - Number of Accidents",
    "Pot Holes": "Pot Holes - Number of Accidents",
    "Steep Grade": "Steep Grade - Number of Accidents",
    "Ongoing Road Works": "Ongoing Road Works/Under Construction - Number of Accidents",
    "Others": "Others - Number of Accidents"
}

road_totals = {}

for feature, column in road_features.items():
    road_totals[feature] = pd.to_numeric(
        a48[column],
        errors="coerce"
    ).sum()

road_summary = pd.DataFrame(
    list(road_totals.items()),
    columns=["Road Feature", "Total Accidents"]
)

road_summary = road_summary.sort_values(
    "Total Accidents",
    ascending=False
)

display(road_summary)
```

```python
plt.figure(figsize=(10, 6))

plt.bar(
    road_summary["Road Feature"],
    road_summary["Total Accidents"]
)

plt.title("Reported Road Accidents by Road Feature")
plt.xlabel("Road Feature")
plt.ylabel("Number of Reported Accidents")

plt.xticks(rotation=35, ha="right")
plt.tight_layout()

plt.show()
```
